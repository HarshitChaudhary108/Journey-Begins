## Q1. What is hierarchical clustering, and how is it different from other clustering techniques?
## Answer 

#### Hierarchical clustering builds a tree-like structure of nested clusters without needing to predefine the number of clusters, unlike K-means or DBSCAN. It’s ideal for visualizing relationships and uncovering subgroups within data.
#### It works in two main ways:
- Agglomerative (Bottom-Up): 
    - Starts with each data point as its own cluster and merges the closest pairs step by step.
- Divisive (Top-Down): 
    - Starts with one big cluster and splits it recursively into smaller clusters.

## Q6. Can hierarchical clustering be used for both numerical and categorical data? If yes, how are the distance metrics different for each type of data?
## Answer 

### For Numerical Data
#### Common Distance Metrics:
- Euclidean Distance: 
    - Measures straight-line distance between points.
- Manhattan Distance: 
    -Measures distance along axes (like city blocks).
- Minkowski Distance: 
    - Generalized form of Euclidean and Manhattan.
### For Categorical Data
#### Common Distance Metrics:
- Hamming Distance: 
    - Counts the number of mismatches between categorical values.
- Jaccard Distance: 
    - Measures dissimilarity based on shared attributes.
- Simple Matching Coefficient (SMC): 
    - Ratio of matching attributes to total attributes.

## Q7. How can you use hierarchical clustering to identify outliers or anomalies in your data?
## Answer 

####  You can use hierarchical clustering to identify outliers by analyzing the dendrogram structure and observing how certain data points behave during the clustering process. Outliers often appear as isolated branches or late-joining points in the hierarchy.
